```python
tensor_1= torch.tensor([[1,2],
                        [4,5],
                        [7,8]])
```

```python
tensor_2 = torch.tensor([[10,11],
                         [12,13],
                         [14,15]])
```

```python
tensor_1.shape ,tensor_2.shape
```

```python
torch.matmul(tensor_1,tensor_2)
```

```python
transpose_of_tensor_2 =tensor_2.T
transpose_of_tensor_2
```

```python
transpose_of_tensor_2.shape
```

```python
torch.matmul(tensor_1,transpose_of_tensor_2)
```

```python
## Here shape of the tensor 1 and 2 are (3,2) and (3,2)
## We Know for multiplication of two matrix of shape (m,n)and (p,q) n=p and the
##Output matrix will be shape (m,q)
```

```python
print(f"Shape of the tensor_1: {tensor_1.shape} and tensor_2: {tensor_2.shape}")
print(f"Shape of the tensor_1: {tensor_1.shape} and shape of the tensor_2 transpose matrix is : {transpose_of_tensor_2.shape}")
print(f"Shape of the output tensor: {torch.matmul(tensor_1,transpose_of_tensor_2).shape}")
```
