```python
#use torch.arange()

zero_to_thousand = torch.arange(start=0,end=1000, step=10)
zero_to_thousand
```

```python
#creagte tensor like
thousand_zeros = torch.zeros_like(input=zero_to_thousand)
thousand_zeros
```

```python
thousand_ones = torch.ones_like(input=zero_to_thousand)
thousand_ones
```
